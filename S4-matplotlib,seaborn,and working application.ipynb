```python
import pandas as pd
import numpy as np
```

```python
sal = pd.read_csv('CaliforniaSalaries .csv')
```

```python
num_cols = ['BasePay','OvertimePay','OtherPay','TotalPay','TotalPayBenefits', 'Year']
for col in num_cols:
    sal[col] = pd.to_numeric(sal[col], errors='coerce')
```

```python
sal
```

```python
import matplotlib.pyplot as plt
```

# 1. Line Plot (Great for showing trends over time)

###to have a clean line plot i will take the mean for each year then plot the graph with the means vs the years

```python
yearly_data = sal.groupby('Year')['TotalPay'].mean()
```

```python
plt.figure(figsize=(6, 4)) 
plt.plot(yearly_data.index, yearly_data.values, color='blue', label='TotalPay Trend') 
plt.title('Basic line plot') 
plt.xlabel('Year') 
plt.ylabel('TotalPay') 
plt.legend()
plt.show() 
```

# 2. Scatter Plot (Great for showing relationship between two numeric variable)

```python
plt.figure(figsize=(6, 4)) 
plt.scatter(sal['BasePay'], sal['OvertimePay'], color='red', alpha=0.7) 
plt.title('Scatter Plot of base pay vs overtime pay') 
plt.xlabel('base pay ($)') 
plt.ylabel('Overtime pay $') 
plt.show()
```

```python

import seaborn as sns
```

```python
plt.figure(figsize=(6, 4)) 
# Compares the spread of Base Pay across different Years 
sns.boxplot(data=sal, x="Year", y="BasePay", palette="Pastel1") 
plt.title("Category Comparison (Box Plot)") 
plt.xlabel("Year") 
plt.ylabel("Base Pay ($)") 
plt.show()
```

#Histogram

```python
plt.figure(figsize=(6, 4)) 

# This automatically counts employees to show the distribution of Total Pay and Benefits 

sns.histplot(data=sal, x="TotalPayBenefits", kde=True, color="purple") 
plt.title("Dataset Distribution (Histogram)") 
plt.xlabel("Total Pay + Benefits ($)") 
plt.ylabel("Number of Employees") 
plt.show()
```
