## Calculating Basic Statistics

### filtering out text columns

```python
numerical_data = sal.select_dtypes(include=['number'])
```

2.calculating individual metrics

```python
print('\n---MEAN---')
print(numerical_data.mean())

print('\n---MEDIAN---')
print(numerical_data.median())

print('\n---MODE---')
print(numerical_data.mode().iloc[0]) 

print('\n---STANDARD DEVIATION---')
print(numerical_data.std()) 

print('\n---CORRELATION MATRIX---')
print(numerical_data.corr()) 
```
